```python
import os
import pandas as pd

folder_path = r"C:\Users\Ved Shankar Patil\Desktop\Projects\zepto_grocery_dashboard"
file_name = "zepto_grocery_data.xlsx"
full_path = os.path.join(folder_path, file_name)

df = pd.read_excel(full_path)

# Verify by showing shape 
print(f"Data loaded successfully from: {full_path}")
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")
display(df.head(20))
```

```python
fat_mapping = {
    "reg": "Regular",
    "LF": "Low Fat",
    "low fat": "Low Fat",
    "Regular": "Regular",
    "Low Fat": "Low Fat",
}
df["Item Fat Content"] = df["Item Fat Content"].map(fat_mapping)
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
display(df.head())
```

```python
with pd.ExcelWriter(
    full_path, engine="openpyxl", mode="a", if_sheet_exists="replace"
) as writer:
  df.to_excel(writer, sheet_name="Cleaned Data", index=False)
```

```python
import os
import pandas as pd
from sqlalchemy import create_engine

# 1. Define your file path
folder_path = r"C:\Users\Ved Shankar Patil\Desktop\Projects\zepto_grocery_dashboard"
file_name = "zepto_grocery_data.xlsx"
full_path = os.path.join(folder_path, file_name)

# 2. Read the cleaned sheet (or the raw sheet if you prefer) from your Excel workbook
df = pd.read_excel(full_path, sheet_name="Cleaned Data")

# 3. Create the connection engine to your PostgreSQL database
# Format: postgresql://username:password@localhost:5432/database_name
engine = create_engine(
    "postgresql://postgres:ABCd12#45@localhost:5432/zepto_grocery_database"
)

# 4. Push the DataFrame into PostgreSQL as a new table
table_name = "zepto_grocery_table"
df.to_sql(name=table_name, con=engine, if_exists="replace", index=False)

print(
    f"Success! Your Excel sheet has been imported into PostgreSQL as table '{table_name}'."
)
```
